A recruitment analytics team wants a compact Python tool to analyze skills requested across job

# Dataset:

```python
job_data = { "Job_ID": [1, 2, 3, 4, 5],
"Role": ["Data Analyst", "Data Scientist", "Business Analyst", "Data Engineer", "ML Engineer"],
"Company": ["ABC Corp", "XyZ Ltd", "DataWorks", "InfraTech", "AI Labs"],
"Skills": [
"Excel SQL Python",
"Python ML Statistics",
"Excel SQL PowerBI",
"Python SQL Spark", "Python ML DeepLearning"
]}
print(job_data)
```

 2) Add New Job Postings (interactive) Ask the user how many job postings they want to add. For each posting, collect the role,
 company, and skills (skills provided as a single string, space-separated or comma-separated).

```python
new_add_job_id=int(input("Enter how many job want to add : "))
id_valu=job_data.get("Job_ID")
add_Role=job_data.get("Role")
add_Company=job_data.get("Company")
add_Skills=job_data.get("Skills")
for i in range(1,new_add_job_id+1):
  valu_1=max(id_valu)+1
  id_valu.append(valu_1)
  add_Role.append(input("Enter role:"))
  add_Company.append(input("Enter company:"))
  add_Skills.append(input("Enter skills (space-separated or comma-separated):"))
  job_data.update({"Job_ID":id_valu})
  job_data.update({"Role":add_Role})
  job_data.update({"Company":add_Company})
  job_data.update({"Skills":add_Skills})
print(job_data)
```

3) Clean & Normalize Skills (text processing)
Make a function called clean_skills() that accepts skills in string format.
* Converts text to lowercase
* Replaces commas with spaces
* Removes extra spaces (normalize whitespace)
* Optionally removes punctuation (you may use basic str.replace() or re module)
Apply clean_skills() to every entry in job_data['Skills'] and overwrite the list with cleaned
versions.

```python
clear_for_Skills = job_data.get("Skills")
def clear_Skills(x):
    x = x.lower()
    x = x.strip()
    x = x.replace(",", " ")
    x = " ".join(x.split())
    return x
after_clean_Skills = []
for clear in clear_for_Skills:
    cleaned = clear_Skills(clear)
    after_clean_Skills.append(cleaned)
job_data.update({"Skills": after_clean_Skills})
print(job_data)
```

4) Skill Extraction & Frequency
* From the cleaned skills list, split each skill string into individual skill tokens.
* Build a consolidated list of all skills that contains every skill token across all postings.
* Use a dictionary or other structure to compute skill_count = {skill: frequency}.
* Print the top 5 most frequent skills (or all if < 5).

```python
count_for_Skills = job_data.get("Skills")
all_Skills=[]
top_skills={}
total_number=[]
after_split_Skills=[]
for Skill in count_for_Skills:
  Skill=Skill.split()
  after_split_Skills.append(Skill)
loa=len(after_split_Skills)
for S1 in range(0,loa):
  for S2 in range(0,len(after_split_Skills[S1])):
    all_Skills.append(after_split_Skills[S1][S2])
print(all_Skills)
for cou in all_Skills:
  total_number.append(all_Skills.count(cou))
print(total_number)
for ts in range(len(total_number)):
  top_skills.update({all_Skills[ts]:total_number[ts]})
sorted_skills = sorted(
    top_skills.items(),
    key=lambda x: x[1],
    reverse=True
)
top_5 = sorted_skills[:5]
print("Top skills by frequency:")
for skill, count in top_5:
    print(f"{skill}: {count}")
```

6) Skill Search Function (functions & conditional logic)
Create a function, find_jobs_by_skill(skill_query), that returns a list of (Job_ID, Role,
Company) for postings containing the skill_query token.
* The function must normalize the skill_query (lowercase, strip spaces)
* Use if/elif/else where appropriate

Example: find_jobs_by_skill('sql') → returns matching jobs.

```python
skill_query = input("Enter the skill: ")
def find_jobs_by_skill(skill_query):
    skill_query = skill_query.lower().strip()
    result = []
    if skill_query == "":
        print("Please enter a skill.")
        return result
    for i in range(len(job_data["Skills"])):
        skills = job_data["Skills"][i].split()
        if skill_query in skills:
            result.append((
                job_data["Job_ID"][i],
                job_data["Role"][i],
                job_data["Company"][i]
            ))
        elif skill_query not in skills:
            continue
        else:
            print("Invalid search.")
    return result
answer = find_jobs_by_skill(skill_query)
if len(answer) == 0:
  print("No matching jobs found.")
else:
  print("Matching Jobs:")
  print(answer)
```

7) Lambda Filtering & List Comprehension
Use a lambda function and filter() (or a list comprehension with a lambda-like expression) to
extract job roles where the skill count (from skill_count) for any skill >= 2 — that is, roles
that include at least one relatively common skill.

Example Output
Roles with at least one common skill: ['Data Analyst', 'Data Scientist', 'ML Engineer']

```python
def check_role(i):
    skills = job_data["Skills"][i].split()
    result = any(
        top_skills.get(skill, 0) >= 2
        for skill in skills
    )
    return result
common_indexes = list(
    filter(
        lambda i: check_role(i),range(len(job_data["Role"]))
    )
)
common_roles = []
for i in common_indexes:
    common_roles.append(job_data["Role"][i])
print("Roles with at least one common skill:")
print(common_roles)
```
